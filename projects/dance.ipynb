```python
from constants import TwelveTET, MIDIinstrument
from converters import comp_to_visual
from analysis import analyze
from structures import Diatonic, MusicComposition, MusicScale, MusicUnit, MusicVoice, MusicTime, PitchRegister

# Berendans
reg = PitchRegister()
time = MusicTime(8,4,4)
# Create units
unit1 = MusicUnit(time,reg)
unit2 = MusicUnit(time,reg)
# Create voices
voice1 = MusicVoice('Melody voice', [unit1], MIDIinstrument.FLUTE)
voice2 = MusicVoice('Accompaniment voice', [unit2], MIDIinstrument.PIANO)
# Create composition
comp = MusicComposition('Berendans',
                   MusicScale(Diatonic.HEPTA, Diatonic.SCALE, TwelveTET.B_FLAT, Diatonic.major_mode),
            [voice1, voice2])

analyze(comp)
comp_to_visual(comp)
```
